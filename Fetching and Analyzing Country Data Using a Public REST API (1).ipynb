```python
import pandas as pd
import requests
#API Endpoint
url = "https://restcountries.com/v3.1/all?fields=name,population,region,area"
#------------------------------------------
#Task 1 — Fetch and Validate the Response
#------------------------------------------
try:
    response = requests.get(url)
    print("Status Code",response.status_code)
    if response.status_code == 200:
        data = response.json()
        print("Total Countries Retrieved:",len(data))
    else:
        print("Error : Failed to fetch data")
except Exception as e:
    print("Exception occured:",str(e))
```

```python
#---------------------------------------------
#Task 2 — Extract and Display Key Fields
#---------------------------------------------
countries_list = []
for country in data:
    name = country.get("name",{}).get("common","N/A")
    population = country.get("population",0)
    region = country.get("region","N/A")
    area = country.get("area",0)
    countries_list.append({
        "Country":name,
        "Population":population,
        "Region":region,
        "Area":area
    })
df = pd.DataFrame(countries_list)
print("\nFirst 10 rows:")
print(df.head(10))
```

```python
#---------------------------------
#Task 3 — Summarize the Data
#---------------------------------

#1.Region with highest total population
region_population = df.groupby("Region")["Population"].sum()
print("\nRegion with highest population:",region_population.idxmax())

#2.Country with largest area
largest_country = df.loc[df["Area"].idxmax(),"Country"]
print("Country with largest area:",largest_country)
```
